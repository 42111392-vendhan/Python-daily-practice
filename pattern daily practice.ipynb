### **Pattern**

```python
def pattern(n):
    for x in range(1,n+1):
        for y in range(x):
            print(x, end="")
        print()
pattern(9)
```

```python
n=5
for i in range(n,0,-1):
        for j in range(i):
            print("&", end="")
        print()
```

```python
a=1
while a<10:
    print(a,end=" ")
    a=a+1
print("\nLoop is done")
```
