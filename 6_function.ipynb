### **function with parameter**

```python
def addition():
    x=10
    y=20
    print("The sum of x and y is:",x+y)
addition()
```

```python
def addition(x,y):
    print("x =",x)
    print("y =",y)
    print("The sum of x and y is:",x+y)
addition(10,20)
```

### **function with default parameter**

```python
def addition(x=0,y=0):
    print("x =",x)
    print("y =",y)
    print("The sum of x and y is:",x+y)
addition()
print()
addition(30,40)
print()
addition(y=60)
```

### **function with return keyword**

```python
def addition(x=0,y=0):
    print("x =",x)
    print("y =",y)
    print("The sum of x and y is:",x+y)

ans=addition(10,20)
print(ans)
```

```python
def addition(x=0,y=0):
    print("x =",x)
    print("y =",y)
    return x+y
ans=addition(10,20)
print("Answer:",ans)
print("Answer 2:",addition(30,40))
```

```python
def add(x, y):
    return x + y
def subtract(x, y):
    return x - y

ans1 = add(10, 5)-subtract(10, 5)
print("Result:", ans1)
```

```python
def add(x, y):
    return x + y
def subtract(x, y):
    return x - y

ans1 = subtract(add(70,30),subtract(50,20))
print("Result:", ans1)
```

```python
def multiply_list(mylist):
    x=1
    for i in mylist:
        x=x*i
    return x

ans2 = multiply_list([1,2,3,4,5,6])
print("Result:", ans2)
```

```python
def myfunc(x,y):
    return x+y
obj = myfunc(10,20)
print("Result:", obj)
```
