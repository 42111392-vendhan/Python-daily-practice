### **1. Syntax error vs runtime error**
*A syntax error occurs when Python code violates Python’s grammar. The program will not start.*

```python
if age >= 18
    print("Eligible")
```

*A runtime error occurs while a syntactically correct program is running.*

```python
print(10 / 0) 
```

### **2. Role of try, except, else, and finally**

* **try: Contains code that may produce an exception.**
* **except: Handles the exception.**
* **else: Executes when no exception occurs.**
* **finally: Always executes, whether an exception occurs or not.**

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
except ValueError:
    print("Enter a valid integer.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
finally:
    print("Program completed.")
```

### **3. Exception raised but not handled**
* *If an exception is not handled, Python stops the program and displays a traceback containing the exception type, message, and location.*

```python
print(10 / 0)
```

### **4. Exception vs BaseException**
* *BaseException is the main parent class of all built-in exceptions.*
* *Exception inherits from BaseException and is the parent of most errors that applications normally handle.*

```python
except Exception as error:
    print(error)
```

*Avoid catching BaseException, because it also catches events such as program exit and keyboard interruption.*

### **5. Use of the raise keyword**
*The raise keyword is used to manually generate an exception.*

```python
age = int(input("Enter your age: "))

if age < 0:
    raise ValueError("Age cannot be negative.")

print("Age:", age)
```

### **6. Creating a custom exception**
*Create a class that inherits from Exception.*

```python
class InsufficientBalanceError(Exception):
    pass

balance = 500
withdrawal = 1000

if withdrawal > balance:
    raise InsufficientBalanceError("Insufficient account balance.")
```

### **7. assert vs raise**
* *assert checks whether a condition is true. If it is false, Python raises an AssertionError.*

```python
age = 20
assert age >= 18, "The person must be at least 18."
```

* *raise manually generates a specific exception.*

```python
if age < 18:
    raise ValueError("The person must be at least 18.")
```

* *assert is mainly used for debugging and internal checks. raise is used for normal error validation. Assertions can also be disabled, so they should not be used for important user-input validation.*

### **8. Multiple except blocks**
*Yes, one try block can have multiple except blocks.*

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
except ValueError:
    print("Invalid input.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
```

### **9. finally without except**
* *A finally block can be used with try even without an except block. The finally block executes before an unhandled exception continues.*

```python
try:
    print(10 / 0)
finally:
    print("This always executes.")
```

* *The message inside finally is printed, and then Python displays the ZeroDivisionError.*

### **10. Checked vs unchecked exceptions**
* *Checked exceptions: Must be handled or declared before compilation in languages such as Java.*
* *Unchecked exceptions: Occur while the program is running and do not require compulsory handling.*
